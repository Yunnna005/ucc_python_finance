```python
#Present Value Calculation

def pres_value_one(CF=2, T=1, DR=5):
    v = CF/(1+DR/100)**T
    return v    
```

```python
print (pres_value_one(100,1,10))
```

```python
#Present Value Calculation with Defaults

def pres_value_two(CF=100, T=1, DR=10):
    v = CF/(1+DR/100)**T
    return v
```

```python
#calculate NPV

cashflows = [-100,10,50,60]
time = [0,1,2,3]
disc_rate = 10

num_cashflows =len(cashflows)
pv_series = [0]*num_cashflows #create a list with num_cashflows element
print(type(pv_series))
print(pv_series)


for x in range(0,num_cashflows): #loop for each cash flow
    pv_series[x] = pres_value_two(cashflows[x],time[x],disc_rate)
print(pv_series)

npv_value=sum(pv_series)
print(npv_value)
```

```python
def npv(cashflows, time, disc_rate):
    num_cashflows =len(cashflows)
    pv_series = [0]*num_cashflows

    for x in range(0,num_cashflows):
        pv_series[x] = pres_value_one(cashflows[x],time[x],disc_rate)

    npv_value=sum(pv_series)
    return(npv_value)

```

```python
CFlows = [-100,10,50,60]
T = [0,1,2,3]
DiscRate = 10

NPV_10 = npv(CFlows,T,DiscRate)
print(NPV_10)

NPV_8 = npv(CFlows,T,8)
print(NPV_8)

NPV_6 = npv(CFlows,T,6)
print(NPV_6)


```

```python
def interp_irr(npv_1, disc_1,npv_2, disc_2):
    irr_est=disc_1+(npv_1/(npv_1-npv_2))*(disc_2-disc_1)
    return irr_est
```

```python
cashflows = [-100000,10000,50000,60000]
time = [0,1,2,3]

disc_rate_low = 6;
disc_rate_high = 10;

npv_low =npv(cashflows,time,disc_rate_low)
npv_high =npv(cashflows,time,disc_rate_high)

irr_est = interp_irr(npv_low, disc_rate_low, npv_high, disc_rate_high)

print(irr_est)
npv_est =npv(cashflows,time,irr_est)
print(npv_est)

irr_inter = interp_irr(npv_est, irr_est, npv_high, disc_rate_high)

print(irr_inter)
npv_est =npv(cashflows,time,irr_inter)
print(npv_est)
```
