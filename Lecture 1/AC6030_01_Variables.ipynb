```python
#variables

list_one = [1,2,3]
list_two=list_one

print(id(list_one))
print(id(list_two))

list_one[0] = 5
print("list_one", list_one)
print("list_two", list_two)

tuple_one = (1,2,3)
tuple_two = tuple_one

print(id(tuple_one))
print(id(tuple_two))

#mutable vs imputable

#tuple_one[0] = 5 #this will give an error
```

```python
#List functions and methods
l_one = [2,4,6,8]

#functions

print("len", len(l_one))
print("sum", sum(l_one))
print("max", max(l_one))
print("min", min(l_one))

#methods
print("count", l_one.count(2))
print("index", l_one.index(6))

print(l_one)
l_one.reverse()
print(l_one)

l_one.append(7)
print(l_one)

l_two = l_one.copy()

print(id(l_one))
print(id(l_two))
```
