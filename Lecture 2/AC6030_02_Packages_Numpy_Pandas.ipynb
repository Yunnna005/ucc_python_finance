```python
#creating arrays
array_one = np.array([1,2,3])
print("Array One: ",array_one)
print("Type: ", type(array_one))

```

```python
#creating a 2-D array

array_two = np.array([[1,2,3],[4,5,6]]) # Note the double [[]]
print(array_two)
#array characteristics
print("Size :", array_two.size)
print("Shape :", array_two.shape) #returns a tuple
print("Dims :", array_two.ndim)
print("Type :", array_two.dtype)
```

```python
a_r = np.arange(12)
print(a_r)

a_rs = a_r.reshape(3,4) # (rows, columns)
print(a_rs)
print()

a_rs2=np.delete(a_rs, 1, axis=1)
print (a_rs2)
```

```python
print(a_rs)
a = a_rs[1,1]
print("a",a)
b = a_rs[1,:] #select row 1,  ":" is not necessary
print("b",b)
c= a_rs[:,2] #select row 1 ":" is necessary
print("c",c)
d = a_rs[0:2,2:4]
print("d")
print(d)

e= np.delete(a_rs, 1, axis=0)
print("e")
print(e)
```

```python
#Methods

print(a_rs)
print(a_rs.sum())
print(a_rs.sum(axis=0))
print(a_rs.sum(axis=1))

#Function

print(np.sum(a_rs))
print(np.sum(a_rs,0))
print(np.sum(a_rs,1))
```
