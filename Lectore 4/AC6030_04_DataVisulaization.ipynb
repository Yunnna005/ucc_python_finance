```python
#This procedure creates two random price series, in one np array for plotting

sim_length = 1000

np.random.seed(sim_length) # Same set of random numbers each time

ri_one = 0.01+np.random.standard_normal(sim_length) # select 100 random numbers from a normal distribution + drift

np.random.seed(sim_length*2) # Same set of random numbers each time
ri_two = 0.01+np.random.standard_normal(sim_length) # select 100 random numbers from a normal distribution + drift

np.random.seed(sim_length*3) # Same set of random numbers each time
market = 0.01+np.random.standard_normal(sim_length) # select 100 random numbers from a normal distribution + drift

price_one = np.cumsum(ri_one+market)
price_two = np.cumsum(ri_two+market)

returns_one = ri_one+market
returns_two = ri_two+market

prices = np.empty((sim_length,2))
prices[:,0] = price_one
prices[:,1] = price_two

returns = np.empty((sim_length,2))
returns[:,0] = returns_one
returns[:,1] = returns_two

print(prices)
print (returns)
```

```python
plt.plot(prices)
```

```python
plt.figure(figsize = (4,2)) 
#plt.plot(prices)

plt.plot(prices[:,0], lw = 1.5, color = 'r', label ='MSFT')
plt.plot(prices[:,1],lw=1.5, color = 'g', label ='IBM')

plt.ylabel('Price')
plt.xlabel('Time')
plt.title('Stock Prices')

#plt.legend(loc=1)
#plt.legend(loc=10)
plt.legend(loc=0)
```
